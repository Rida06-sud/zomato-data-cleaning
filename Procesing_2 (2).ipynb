```python
def fix_coord(x, lo, hi):
    if pd.isna(x):
        return np.nan
    sign = -1 if x.startswith("-") else 1
    digits = x.replace("-", "").replace(".", "")
    if int(digits) == 0:
        return np.nan
    for k in (1, 2, 3):
        if len(digits) > k:
            val = sign * float(digits[:k] + "." + digits[k:])
            if lo <= val <= hi:
                return val
    return np.nan

batas = {
    1:   ((68, 98),    (6, 37)),
    14:  ((112, 154),  (-44, -10)),
    30:  ((-75, -34),  (-34, -10)),
    37:  ((-141, -52), (41, 84)),
    94:  ((95, 141),   (-11, 6)),
    148: ((166, 179),  (-47, -34)),
    162: ((116, 127),  (4, 21)),
    166: ((50, 52),    (24, 27)),
    184: ((103, 105),  (1, 2)),
    189: ((16, 33),    (-35, -22)),
    191: ((79, 82),    (5, 10)),
    208: ((26, 45),    (36, 42)),
    214: ((51, 57),    (22, 26.5)),
    215: ((-8, 2),     (49, 61)),
    216: ((-125, -66), (24, 50)),
}

df["Longitude"] = df.apply(lambda r: fix_coord(r["Longitude"], *batas[r["Country Code"]][0]), axis=1)
df["Latitude"]  = df.apply(lambda r: fix_coord(r["Latitude"],  *batas[r["Country Code"]][1]), axis=1)
```

Fungsi fix_coord(x, lo, hi):
Simpan tanda negatif, lalu ambil semua digit tanpa titik/minus.
Kalau semuanya nol, kembalikan NaN.
Coba sisipkan titik desimal setelah digit ke-1, ke-2, atau ke-3.
Terima hasil pertama yang masuk di rentang lo–hi (batas wajar koordinat negara tersebut), kalau tidak ada yang cocok jadi NaN.
Dictionary batas: rentang longitude/latitude yang valid per negara, jadi posisi titik desimal bisa ditebak dengan benar.
df.apply(lambda r: ..., axis=1): terapkan fungsi ke tiap baris sesuai negaranya.

```python
# isi data kosong
df["Cuisines"] = df["Cuisines"].fillna("Unknown")

# biaya: median per Country + Price range, lalu per Country
df["Average Cost for two"] = df.groupby(["Country", "Price range"])["Average Cost for two"] \
                               .transform(lambda s: s.fillna(s.median()))
df["Average Cost for two"] = df.groupby("Country")["Average Cost for two"] \
                               .transform(lambda s: s.fillna(s.median()))

# koordinat: median per City, lalu per Country
for c in ["Longitude", "Latitude"]:
    df[c] = df.groupby("City")[c].transform(lambda s: s.fillna(s.median()))
    df[c] = df.groupby("Country")[c].transform(lambda s: s.fillna(s.median()))

# cek sisa kosong (harusnya cuma Aggregate rating)
df.isna().sum()[df.isna().sum() > 0]
```
